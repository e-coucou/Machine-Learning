```python
@dataclass
class ModelConfig:
    batch_size: int = 64
    block_size: int = 256
    max_iters: int = 5000
    eval_interval: int = 500
    eval_iters: int = 200
    learning_rate: float = 3e-4
    n_embd: int = 384
    num_heads: int = 6
    n_layers: int = 6
    dropout: float = 0.2

config = { 'n_embd': 384,
            'num_heads': 3,
            'n_layers': 3,
            'block_size': 32,
            'dropout': 0.2,      
        }
parameters = {
    'batch_size': 32,
    'max_iters': 2000,
    'eval_interval': 500,
    'eval_iters': 200,
    'learning_rate': 1e-4
}
```

```python
decoder = at.Model(token_.ids, token_, **config, **parameters, fileout='data/test_2.pth')
decoder.get_model_size()
```

```python
decoder.train()
decoder.plot_losses()
```

```python
import regex as rex

mess = "bonjour y a-t'il un probleme aujourd'hui puiqu'il y a des sauf-conduits pour les 3256"
GPT_EP  = r"""(?i:[lcdtmnsj]|qu)'|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
GPT4_SP  = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
TEST = r"""(?i:[lcdtmnsjh]|qu)'|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}"""
pattern = rex.compile(TEST)
pattern_final = rex.compile(GPT_EP)
token = rex.findall(pattern, mess)
token_final = rex.findall(pattern_final, mess)
print(token)
print(token_final)
```
