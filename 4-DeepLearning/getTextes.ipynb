```python
# Script pour télécharger plusieurs œuvres littéraires françaises depuis Project Gutenberg
"""
# Les Misérables
url = "https://www.gutenberg.org/files/17489/17489-0.txt"
urllib.request.urlretrieve(url, "les_miserables.txt")

# Notre-Dame de Paris
url = "https://www.gutenberg.org/files/2610/2610-0.txt"
urllib.request.urlretrieve(url, "notre_dame_paris.txt")
Alexandre Dumas
python# Les Trois Mousquetaires
url = "https://www.gutenberg.org/files/1257/1257-0.txt"
urllib.request.urlretrieve(url, "trois_mousquetaires.txt")

# Le Comte de Monte-Cristo
url = "https://www.gutenberg.org/files/17989/17989-0.txt"
urllib.request.urlretrieve(url, "monte_cristo.txt")
Gustave Flaubert
python# Madame Bovary
url = "https://www.gutenberg.org/files/14155/14155-0.txt"
urllib.request.urlretrieve(url, "madame_bovary.txt")
Jules Verne
python# Vingt mille lieues sous les mers
url = "https://www.gutenberg.org/files/5097/5097-0.txt"
urllib.request.urlretrieve(url, "vingt_mille_lieues.txt")

# Le Tour du monde en 80 jours
url = "https://www.gutenberg.org/files/800/800-0.txt"
urllib.request.urlretrieve(url, "tour_du_monde_80_jours.txt")
Honoré de Balzac
python# Le Père Goriot
url = "https://www.gutenberg.org/files/1237/1237-0.txt"
urllib.request.urlretrieve(url, "pere_goriot.txt")
Émile Zola
python# Germinal
url = "https://www.gutenberg.org/files/5711/5711-0.txt"
urllib.request.urlretrieve(url, "germinal.txt")
2. Script pour télécharger plusieurs œuvres
pythonimport urllib.request
import time
"""
import os, sys, urllib.request, time

def download_book(url, filename):
    """Télécharge un livre depuis Project Gutenberg"""
#    filename += "data/"
    try:
        print(f"Téléchargement de {filename}...")
        urllib.request.urlretrieve(url, f"data/{filename}")
        print(f"✅ {filename} téléchargé")
        time.sleep(1)  # Pause pour ne pas surcharger le serveur
        return True
    except Exception as e:
        print(f"❌ Erreur pour {filename}: {e}")
        return False

# Dictionnaire d'œuvres françaises
french_books = {
    # Victor Hugo
    "les_miserables.txt": "https://www.gutenberg.org/cache/epub/17489/pg17489.txt",
    # Alexandre Dumas
    "trois_mousquetaires.txt": "https://www.gutenberg.org/cache/epub/13951/pg13951.txt",
    "monte_cristo_tome1.txt": "https://www.gutenberg.org/cache/epub/17989/pg17989.txt",
    "monte_cristo_tome2.txt": "https://www.gutenberg.org/cache/epub/17990/pg17990.txt",
    # Gustave Flaubert
    "madame_bovary.txt": "https://www.gutenberg.org/cache/epub/14155/pg14155.txt",
    # Émile Zola
    "germinal.txt": "https://www.gutenberg.org/cache/epub/5711/pg5711.txt",
    # Jules Verne
    "vingt_mille_lieues.txt": "https://www.gutenberg.org/cache/epub/5097/pg5097.txt",
    "tour_du_monde_80_jours.txt": "https://www.gutenberg.org/cache/epub/800/pg800.txt",
    # Molière
    # Guy de Maupassant
    "bel_ami.txt": "https://www.gutenberg.org/cache/epub/5711/pg5711.txt",
    # Molière
    "moliere_1":"https://www.gutenberg.org/cache/epub/40086/pg40086.txt",
    "moliere_2":"https://www.gutenberg.org/cache/epub/43535/pg43535.txt",
    "moliere_3":"https://www.gutenberg.org/cache/epub/50173/pg50173.txt",
}

# Télécharger tous les livres
for filename, url in french_books.items():
    download_book(url, filename)

print("\n✅ Téléchargements terminés!")
```

```python
def combine_books(input_files, output_file):
    """Combine plusieurs livres en un seul corpus"""
    combined = []
    
    for filename in input_files:
        filepath = f"data/{filename}"
        if os.path.exists(filepath):
            with open(filepath, 'r', encoding='utf-8') as f:
                content = f.read().strip()
                combined.append(content)
                print(f"Ajouté: {filename} ({len(content):,} caractères)")
        else:
            print(f"⚠️ Fichier non trouvé: {filename}")
    
    # Joindre avec séparateurs
    full_text = "\n\n" + "="*80 + "\n\n".join(combined)
    
    # Sauvegarder
    with open(f"data/{output_file}", 'w', encoding='utf-8') as f:
        f.write(full_text)
    
    print(f"\n✅ Corpus combiné: {output_file}")
    print(f"   Total: {len(full_text):,} caractères")

# Créer un corpus d'entraînement
combine_books(
    [
        "bel_ami.txt",
        "germinal.txt",
        "les_miserables.txt",
        "madame_bovary.txt",
        "moliere_1.txt",
        "moliere_2.txt",
        "moliere_3.txt",
        "les_miserables.txt",
        "monte_cristo_tome1.txt",
        "monte_cristo_tome2.txt",
        "tour_du_monde_80_jours.txt",
        "trois_mousquetaires.txt",
        "vingt_mille_lieues.txt"],
    "corpus_francais.txt"
)
```
