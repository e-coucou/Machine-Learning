```python
def combine_books(input_files, output_file):
    """Combine plusieurs livres en un seul corpus"""
    combined = []
    
    for filename in input_files:
        filepath = f"data/{filename}"
        if os.path.exists(filepath):
            with open(filepath, 'r', encoding='utf-8') as f:
                content = f.read().strip()
                combined.append(content)
                print(f"Ajouté: {filename} ({len(content):,} caractères)")
        else:
            print(f"⚠️ Fichier non trouvé: {filename}")
    
    # Joindre avec séparateurs
    full_text = "\n\n" + "="*80 + "\n\n".join(combined)
    
    # Sauvegarder
    with open(f"data/{output_file}", 'w', encoding='utf-8') as f:
        f.write(full_text)
    
    print(f"\n✅ Corpus combiné: {output_file}")
    print(f"   Total: {len(full_text):,} caractères")

# Créer un corpus d'entraînement
combine_books(
    [
        "bel_ami.txt",
        "germinal.txt",
        "les_miserables.txt",
        "madame_bovary.txt",
        "moliere_1.txt",
        "moliere_2.txt",
        "moliere_3.txt",
        "les_miserables.txt",
        "monte_cristo_tome1.txt",
        "monte_cristo_tome2.txt",
        "tour_du_monde_80_jours.txt",
        "trois_mousquetaires.txt",
        "vingt_mille_lieues.txt"],
    "corpus_francais.txt"
)
```

```python
import os
import json
import glob
import random

# -----------------------------
# Paramètres
# -----------------------------
WIKI_DIR = "wiki_fr"          # dossier WikiExtractor
OUTPUT_FILE = "wiki_fr_dataset_llm.jsonl"
INSTRUCTION = "Résume le texte suivant en français."
MAX_PARAGRAPH_LEN = 500       # caractères max d'un paragraphe à traiter

# -----------------------------
# Collecte des fichiers txt extraits
# -----------------------------
files = glob.glob(os.path.join(WIKI_DIR, "**", "*.txt"), recursive=True)
print(f"Nombre de fichiers trouvés : {len(files)}")

dataset = []

# -----------------------------
# Boucle de génération
# -----------------------------
for file_path in files:
    with open(file_path, "r", encoding="utf-8") as f:
        paragraphs = f.read().split("\n\n")
        for para in paragraphs:
            para = para.strip()
            if len(para) < 100 or len(para) > MAX_PARAGRAPH_LEN:
                continue  # filtrer les paragraphes trop courts ou trop longs

            # Génération du résumé via LLM
            try:
                summary = summarizer(para, max_length=150, min_length=30, do_sample=False)[0]["summary_text"]
            except Exception as e:
                print(f"Erreur pour un paragraphe : {e}")
                continue

            dataset.append({
                "instruction": INSTRUCTION,
                "input": para,
                "output": summary
            })

# Shuffle
random.shuffle(dataset)

# Écriture JSONL
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    for entry in dataset:
        f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print(f"Dataset créé : {OUTPUT_FILE}, {len(dataset)} exemples")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

model_name = "google/mt5-small" #"plguillou/t5-small-fr-sum-cnndm"

tokenizer = AutoTokenizer.from_pretrained(model_name,  use_fast=False)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

summarizer = pipeline("summarization", model=model, tokenizer=tokenizer, framework="pt")

# Test rapide
text = "Le Chat est un animal domestique très populaire en France. Il est apprécié pour sa compagnie et sa douceur."
print(summarizer(text, max_length=50, min_length=10)[0]['summary_text'])
```
